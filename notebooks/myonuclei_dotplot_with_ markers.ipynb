```python
from pathlib import Path
import anndata as ad
import matplotlib.pyplot as plt
import scanpy as sc
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
myonuclei = ["Myonuclei_Trim63", "Myonuclei_NMJ", "Myonuclei_MTJ", "Myonuclei_IIx_IIb", "Myonuclei_IIx_IIa", "Myonuclei_IIx", "Myonuclei_IIb"]
myonuclei_marker_genes_list = ["Ache", "Col22a1", "Etv5", "Myh1", "Myh2", "Myh4", "Ttn", "Trim63"]
```

```python
#create the dot plots
INPUT_DIR = Path('/home/stefano/Documents/Spatial-Transcriptomic/results/rctd_result/weights')
OUTPUT_DIR = Path('/home/stefano/Documents/Spatial-Transcriptomic/results/rctd_result/plots')

h5ad_files = list(INPUT_DIR.glob("*.h5ad"))

for h5ad_path in h5ad_files:
    adata = ad.read_h5ad(h5ad_path)
    block_id = h5ad_path.stem

    dp_myonuclei = dp_myonuclei = sc.pl.dotplot(
                adata[adata.obs["fiber_type"].isin(myonuclei)],
                var_names = myonuclei_marker_genes_list,
                groupby = "fiber_type",
                use_raw = False, #since we normalize
                standard_scale = 'var',
                colorbar_title='Mean expression\n(standard scaled)',
                title = f"Myonuclei marker expression - {block_id} ",
                show=False
        )

    #save the dot plot
    plot_filename =OUTPUT_DIR/ f"{block_id}_dotplot_myonuclei_markers.png"
    plt.savefig(plot_filename, bbox_inches='tight', dpi=300)
    plt.close()
    print(f"{block_id} saved")
```
